#### WAP py, for a food court where first it displays all the food items and their rates available. The useer has to pass the item code
#### and its quantity .
#### It will ask until the user press X in the item code and quantity 0

```python
# Food Court Billing System

print("********** FOOD MENU **********")
print("Code\tItem\t\tPrice")
print("101\tBurger\t\t120")
print("102\tPizza\t\t250")
print("103\tSandwich\t100")
print("104\tFries\t\t90")
print("105\tCold Drink\t50")

# Variables
grand_total = 0

# Lists to store bill details
item_list = []
price_list = []
qty_list = []
amount_list = []

while True:
    code = input("\nEnter Item Code (X to Exit): ").upper()

    if code == "X":
        qty = int(input("Enter Quantity (0 to Exit): "))
        if qty == 0:
            break

    qty = int(input("Enter Quantity: "))

    if code == "101":
        item = "Burger"
        price = 120
    elif code == "102":
        item = "Pizza"
        price = 250
    elif code == "103":
        item = "Sandwich"
        price = 100
    elif code == "104":
        item = "Fries"
        price = 90
    elif code == "105":
        item = "Cold Drink"
        price = 50
    else:
        print("Invalid Item Code!")
        continue

    amount = price * qty
    grand_total += amount

    # Store details
    item_list.append(item)
    price_list.append(price)
    qty_list.append(qty)
    amount_list.append(amount)

print("\n\n*************** BILL ***************")
print("Item\t\tPrice\tQty\tAmount")

for i in range(len(item_list)):
    print(item_list[i], "\t\t", price_list[i], "\t", qty_list[i], "\t", amount_list[i])

print("-----------------------------------------")
print("Grand Total =", grand_total)
print("-----------------------------------------")
print("Thank You! Visit Again.")
```

```python
from datetime import date

def findAge(dob):
    today = date.today()
    age = today.year - dob.year

    if (today.month, today.day) < (dob.month, dob.day):
        age -= 1

    return age

day, month, year = map(int, input("Enter DOB (DD MM YYYY): ").split())
dob = date(year, month, day)

print("Age =", findAge(dob))
```

```python
# WAP to calculate factorial using a function

def factorial(n):
    pro = 1
    for i in range(1, n + 1):
        pro = pro * i
    return pro

n = int(input("Enter n: "))
print("n =", n, "and n! =", factorial(n))
```
