```python
#warehouse management
warehouse = {
    "Laptop": 10,
    "Mobile": 20,
    "Keyboard": 15,
    "Mouse": 25
}

print("Warehouse Inventory:")
print(warehouse)

# Add a new product
warehouse["Headphones"] = 12

# Update stock
warehouse["Laptop"] = 15

# Remove a product
del warehouse["Mouse"]

print("\nUpdated Warehouse Inventory:")
print(warehouse)

# Search for a product
product = input("\nEnter product name: ")

if product in warehouse:
    print("Available Stock:", warehouse[product])
else:
    print("Product not found")
```

```python
#student marks analysis
import pandas as pd
marks = [85, 72, 90, 65, 78]

print("Student Marks:", marks)

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print(pd.Series(marks).describe())
```

```python
#file handling
# Writing student record to a file

file = open("student.txt", "w")

file.write("Student Name: Vaibhav\n")
file.write("Roll Number: 90\n")
file.write("Marks: 100\n")

file.close()

# Reading student record

file = open("student.txt", "r")

print("Student Record:")
print(file.read())

file.close()
```

```python
class Student:
    def accept(self):
        self.roll = int(input("Enter roll number: "))
        self.name = input("Enter name: ")
        self.marks = int(input("Enter marks: "))

    def display(self):
        print("Roll Number:", self.roll)
        print("Name:", self.name)
        print("Marks:", self.marks)

    def grade(self):
        if self.marks >= 40:
            print("Grade: Pass")
        else:
            print("Grade: Fail")


s = Student()

s.accept()
s.display()
s.grade()
```
