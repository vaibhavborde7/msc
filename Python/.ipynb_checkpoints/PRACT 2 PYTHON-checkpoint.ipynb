```python
x = int(input())
y = int(input())
z = int(input())
n = int(input())

result = [[i, j, k]
          for i in range(x + 1)
          for j in range(y + 1)
          for k in range(z + 1)
          if i + j + k != n]

print(result)
```

```python
product = ("Laptop", 50000, "Electronics")

print("Product Name:", product[0])
print("Price:", product[1])
print("Category:", product[2])

print("Length of tuple:", len(product))

new_product = product + ("Available",)

print("Updated Tuple:", new_product)
```

```python
customers1 = {101, 102, 103, 104}
customers2 = {103, 104, 105, 106}

print("Union:", customers1 | customers2)
print("Intersection:", customers1 & customers2)
print("Difference:", customers1 - customers2)
print("Symmetric Difference:", customers1 ^ customers2)
```

```python
customers = {
    101: "Rahul",
    102: "Priya",
    103: "Amit"
}

print("Customer Database:")
print(customers)

print("Customer 101:", customers[101])

customers[104] = "Sneha"

customers[102] = "Pooja"

del customers[103]

print("Updated Database:")
print(customers)
```

```python
text = input("Enter a string: ")

print("Original String:", text)
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Length:", len(text))
print("Reversed String:", text[::-1])

if text.lower() == text[::-1].lower():
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b

    print("Result:", result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero")

except ValueError:
    print("Error: Please enter numbers only")
```
