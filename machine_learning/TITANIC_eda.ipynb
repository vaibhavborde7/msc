```python
import pandas as pd

'''csv reading'''
data = pd.read_csv("Titanic-Dataset_check.csv")

df = pd.DataFrame(data)

# 1. Display first 5 rows
print("First 5 Rows:")
print(df.head())

# 2. Display last 5 rows
print("\nLast 5 Rows:")
print(df.tail())
```

```python
# 3. Dataset information
print("\nDataset Information:")
df.info()

# 4. Statistical Summary
print("\nStatistical Summary:")
print(df.describe())

# 5. Shape of dataset
print("\nShape:")
print(df.shape)

# 6. Column Names
print("\nColumns:")
print(df.columns)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Step 3: Replace missing prices with mean age
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
# Step 4: Check duplicate records
print("\nDuplicate Records:")
print(df[df.duplicated()])
```

```python
# 8. Duplicate Rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python


# 9. Unique Values
print("\nUnique Values:")
print(df.nunique())

# 10. Survival Count
print("\nSurvival Count:")
print(df["Survived"].value_counts())

# 11. Passenger Class Count
print("\nPassenger Class:")
print(df["Pclass"].value_counts())

# 12. Gender Count
print("\nGender Count:")
print(df["Sex"].value_counts())

# 13. Embarked Port Count
print("\nEmbarked Count:")
print(df["Embarked"].value_counts())

# 14. Average Age
print("\nAverage Age:")
print(df["Age"].mean())

# 15. Average Fare
print("\nAverage Fare:")
print(df["Fare"].mean())
```

```python
#Max Age
print(df["Age"].max())

# Minimum Age
print(df["Age"].min())

#Passengers older than 45
print("Passengers older than 70")
age70 = df[df["Age"] > 70]
print(age70)
```
