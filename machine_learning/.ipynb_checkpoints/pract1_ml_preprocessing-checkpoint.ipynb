```python
'''handling missing values'''
import numpy as np

arr = np.array([10, 20, np.nan, 40, 50])

# Calculate mean (ignoring NaN)
mean_value = np.nanmean(arr)

# Replace NaN with mean
arr[np.isnan(arr)] = mean_value

print(arr)
```

```python
''' Statistical operations'''

print("mean : ", np.mean(data))
print("median : ", np.median(data))
print("Standard Deviation : " , np.std(data))
print("Maximum : ", np.max(data))
print("Minimum : ", np.min(data))
```

```python
import numpy as np

salary = np.array([25000, 30000, 35000, 45000])

print(salary[salary > 30000])
```

```python

# Step 1: Create salary data (with a missing value)
salary = np.array([25000, 30000, 35000, np.nan, 45000])

print("Original Salary Data:")
print(salary)

# Step 2: Replace missing value with average salary
salary[np.isnan(salary)] = np.nanmean(salary)

print("\nAfter Replacing Missing Value:")
print(salary)

# Step 3: Compute average salary
average = np.mean(salary)
print("\nAverage Salary =", average)

# Step 4: Normalize salary (Min-Max Normalization)
normalized_salary = (salary - salary.min()) / (salary.max() - salary.min())

print("\nNormalized Salary:")
print(normalized_salary)

# Step 5: Filter salaries greater than ₹30,000
high_salary = salary[salary > 30000]

print("\nEmployees earning more than ₹30,000:")
print(high_salary)
```

```python
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    "Salary": [25000, 30000, np.nan, 35000, 45000]
})

print("Original Data:")
print(df)

# Replace missing value with mean
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("\nAfter Replacing Missing Values:")
print(df)
```

```python
print(df.isnull())
```
