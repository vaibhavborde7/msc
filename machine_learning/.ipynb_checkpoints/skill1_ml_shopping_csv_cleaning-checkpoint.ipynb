### ML Skill 1: Shopping Data Cleaning

```python
import pandas as pd

'''csv reading'''
data = pd.read_csv("skill1_shopping_data.csv")

print(data)
df = pd.DataFrame(data)
```

```python
# Step 2: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Step 3: Replace missing prices with mean price
df["price"] = df["price"].fillna(df["price"].mean())
```

```python
# Step 4: Check duplicate records
print("\nDuplicate Records:")
print(df[df.duplicated()])
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Step 5: Correct incorrect entries
# Fix spelling mistake in Product column
df["product"] = df["product"].replace("Moblie", "Mobile")
```

```python
# Step 6: Business Insights

# Average Price
print("\nAverage Product Price:")
print(df["price"].mean())

# Total Sales
print("\nTotal Sales:")
print(df["price"].sum())

# Number of customers per city
print("\nCustomers by City:")
print(df["city"].value_counts())

# Product-wise Average Price
print("\nAverage Price by Product:")
print(df.groupby("product")["price"].mean())
```

```python
# Step 7: Identify Laptop Customers
print("\nLaptop Customers:")
laptop_customers = df[df["product"] == "Laptop"]
print(laptop_customers)
```

```python
# Step 8: Final Cleaned Dataset
print("\nCleaned Dataset")
print(df)
```
